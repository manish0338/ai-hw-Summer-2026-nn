```python
# ============================================================================
#  MNIST: MLP vs CNN vs Transformer
# ============================================================================
!pip -q install torch torchvision --upgrade

import os, time, json, math, random
import numpy as np
import torch, torch.nn as nn, torch.nn.functional as F
from torch.utils.data import DataLoader
from torchvision import datasets, transforms

SEED = 42
random.seed(SEED); np.random.seed(SEED)
torch.manual_seed(SEED); torch.cuda.manual_seed_all(SEED)

DEVICE  = "cuda" if torch.cuda.is_available() else "cpu"
EPOCHS  = 8
BATCH   = 128
OUTDIR  = "results"; os.makedirs(OUTDIR, exist_ok=True)
print("device:", DEVICE)

MEAN, STD = (0.1307,), (0.3081,)

train_tf = transforms.Compose([
    transforms.RandomAffine(degrees=10, translate=(0.1, 0.1),
                            scale=(0.9, 1.1), shear=5),
    transforms.ToTensor(),
    transforms.Normalize(MEAN, STD),
    transforms.RandomErasing(p=0.25, scale=(0.02, 0.1)),
])
test_tf = transforms.Compose([transforms.ToTensor(), transforms.Normalize(MEAN, STD)])

train_ds = datasets.MNIST("./data", train=True,  download=True, transform=train_tf)
test_ds  = datasets.MNIST("./data", train=False, download=True, transform=test_tf)

train_ld = DataLoader(train_ds, batch_size=BATCH, shuffle=True,  num_workers=2, pin_memory=True)
test_ld  = DataLoader(test_ds,  batch_size=512,   shuffle=False, num_workers=2, pin_memory=True)
print(f"train: {len(train_ds)}  test: {len(test_ds)}")

# 1) shallow MLP
class MLP(nn.Module):
    def __init__(self, hidden=256, ncls=10):
        super().__init__()
        self.net = nn.Sequential(
            nn.Flatten(),
            nn.Linear(28*28, hidden), nn.ReLU(), nn.Dropout(0.2),
            nn.Linear(hidden, ncls))
    def forward(self, x): return self.net(x)

# 2) small CNN
class CNN(nn.Module):
    def __init__(self, ncls=10):
        super().__init__()
        self.features = nn.Sequential(
            nn.Conv2d(1, 32, 3, padding=1), nn.BatchNorm2d(32), nn.ReLU(),
            nn.Conv2d(32, 32, 3, padding=1), nn.BatchNorm2d(32), nn.ReLU(),
            nn.MaxPool2d(2), nn.Dropout(0.25),                       # 14x14
            nn.Conv2d(32, 64, 3, padding=1), nn.BatchNorm2d(64), nn.ReLU(),
            nn.Conv2d(64, 64, 3, padding=1), nn.BatchNorm2d(64), nn.ReLU(),
            nn.MaxPool2d(2), nn.Dropout(0.25),                       # 7x7
        )
        self.head = nn.Sequential(nn.Flatten(), nn.Linear(64*7*7, 128),
                                  nn.ReLU(), nn.Dropout(0.5), nn.Linear(128, ncls))
    def forward(self, x): return self.head(self.features(x))

# 3) Transformer encoder
class ViT(nn.Module):
    """28x28 -> 7x7 patches of size 4x4 = 49 tokens + CLS token."""
    def __init__(self, patch=4, dim=128, depth=4, heads=4, ncls=10):
        super().__init__()
        self.patch_embed = nn.Conv2d(1, dim, kernel_size=patch, stride=patch)
        n_tok = (28 // patch) ** 2
        self.cls = nn.Parameter(torch.zeros(1, 1, dim))
        self.pos = nn.Parameter(torch.randn(1, n_tok + 1, dim) * 0.02)
        layer = nn.TransformerEncoderLayer(d_model=dim, nhead=heads,
                                           dim_feedforward=dim*4, dropout=0.1,
                                           activation="gelu", batch_first=True,
                                           norm_first=True)
        self.encoder = nn.TransformerEncoder(layer, num_layers=depth)
        self.norm = nn.LayerNorm(dim)
        self.fc   = nn.Linear(dim, ncls)
    def forward(self, x):
        x = self.patch_embed(x).flatten(2).transpose(1, 2)      # B,N,dim
        x = torch.cat([self.cls.expand(x.size(0), -1, -1), x], 1) + self.pos
        x = self.encoder(x)
        return self.fc(self.norm(x[:, 0]))

#  train / eval
@torch.no_grad()
def evaluate(model, loader):
    model.eval(); correct = loss_sum = n = 0
    per_cls_c = torch.zeros(10); per_cls_n = torch.zeros(10)
    for x, y in loader:
        x, y = x.to(DEVICE), y.to(DEVICE)
        out = model(x)
        loss_sum += F.cross_entropy(out, y, reduction="sum").item()
        p = out.argmax(1); correct += (p == y).sum().item(); n += y.numel()
        for c in range(10):
            m = (y == c); per_cls_n[c] += m.sum().item()
            per_cls_c[c] += (p[m] == c).sum().item()
    return correct / n, loss_sum / n, (per_cls_c / per_cls_n.clamp(min=1)).tolist()

def run(name, model):
    model = model.to(DEVICE)
    nparam = sum(p.numel() for p in model.parameters() if p.requires_grad)
    opt = torch.optim.AdamW(model.parameters(), lr=3e-3 if name != "Transformer" else 1e-3,
                            weight_decay=5e-4)
    sched = torch.optim.lr_scheduler.OneCycleLR(opt, max_lr=opt.param_groups[0]["lr"],
                                                total_steps=EPOCHS*len(train_ld))
    hist, best, t0 = [], 0.0, time.time()
    print(f"\n=== {name} | params: {nparam:,} ===")
    for ep in range(1, EPOCHS + 1):
        model.train(); tl = tc = tn = 0
        for x, y in train_ld:
            x, y = x.to(DEVICE, non_blocking=True), y.to(DEVICE, non_blocking=True)
            opt.zero_grad(set_to_none=True)
            out = model(x); loss = F.cross_entropy(out, y, label_smoothing=0.05)
            loss.backward(); nn.utils.clip_grad_norm_(model.parameters(), 1.0)
            opt.step(); sched.step()
            tl += loss.item()*y.numel(); tc += (out.argmax(1) == y).sum().item(); tn += y.numel()
        te_acc, te_loss, per_cls = evaluate(model, test_ld)
        best = max(best, te_acc)
        hist.append(dict(epoch=ep, train_loss=tl/tn, train_acc=tc/tn,
                         test_loss=te_loss, test_acc=te_acc))
        print(f"ep {ep:02d}/{EPOCHS}  train_loss {tl/tn:.4f} acc {tc/tn*100:5.2f}% | "
              f"TEST loss {te_loss:.4f} acc {te_acc*100:5.2f}%")
    torch.save(model.state_dict(), f"{OUTDIR}/{name.lower()}_mnist.pt")
    return dict(model=name, params=nparam, epochs=EPOCHS,
                final_test_acc=hist[-1]["test_acc"], best_test_acc=best,
                test_error_pct=100*(1-best), minutes=(time.time()-t0)/60,
                per_class_acc={str(i): round(a, 4) for i, a in enumerate(per_cls)},
                history=hist)

results = [run("MLP", MLP()), run("CNN", CNN()), run("Transformer", ViT())]

# ------------------------------ report --------------------------------------
print("\n================= FINAL TEST RESULTS (MNIST test set, 10k) =================")
hdr = f"{'Model':<12}{'Params':>10}{'Best test acc':>15}{'Test error':>12}{'Time,min':>10}"
print(hdr); print("-"*len(hdr))
lines = ["| Model | Params | Best test acc | Test error | Train time (min) |",
         "|---|---|---|---|---|"]
for r in results:
    print(f"{r['model']:<12}{r['params']:>10,}{r['best_test_acc']*100:>14.2f}%"
          f"{r['test_error_pct']:>11.2f}%{r['minutes']:>10.1f}")
    lines.append(f"| {r['model']} | {r['params']:,} | {r['best_test_acc']*100:.2f}% | "
                 f"{r['test_error_pct']:.2f}% | {r['minutes']:.1f} |")

with open(f"{OUTDIR}/results.json", "w") as f: json.dump(results, f, indent=2)
with open(f"{OUTDIR}/RESULTS.md", "w") as f:
    f.write("# MNIST results (trained on train split, evaluated on test split)\n\n"
            f"Epochs: {EPOCHS}, batch: {BATCH}, optimizer: AdamW + OneCycle, "
            "augment: RandomAffine(rot10, trans0.1, scale0.9-1.1, shear5) + RandomErasing\n\n"
            + "\n".join(lines) + "\n")
print(f"\nSaved: {OUTDIR}/results.json, {OUTDIR}/RESULTS.md, {OUTDIR}/*.pt")

# ------------------------------ plots ---------------------------------------
import matplotlib.pyplot as plt
fig, ax = plt.subplots(1, 2, figsize=(12, 4))
for r in results:
    ep = [h["epoch"] for h in r["history"]]
    ax[0].plot(ep, [h["test_acc"]*100 for h in r["history"]], marker="o", label=r["model"])
    ax[1].plot(ep, [h["test_loss"] for h in r["history"]], marker="o", label=r["model"])
ax[0].set_title("Test accuracy, %"); ax[1].set_title("Test loss")
for a in ax: a.set_xlabel("epoch"); a.grid(alpha=.3); a.legend()
plt.tight_layout(); plt.savefig(f"{OUTDIR}/curves.png", dpi=140); plt.show()
```
